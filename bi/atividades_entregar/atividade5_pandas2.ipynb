```python
# ----------------FIM questão 9 ----------------

data_loja1 = {
    'Produto': pd.Series(['Smartphone', 'Tablet', 'Notebook', 'Impressora', 'Roteador'], index=[1, 2, 3, 4, 5]),
    'Estoque': pd.Series([12, 8, 5, 10, 20], index=[1, 2, 3, 4, 5])
}
df_loja1 = pd.DataFrame(data_loja1)

data_loja2 = {
    'Produto': pd.Series(['Cadeira Gamer', 'Mesa Office', 'Luminária', 'Nobreak', 'Monitor Curvo'], index=[1, 2, 3, 4, 5]),
    'Estoque': pd.Series([7, 4, 15, 6, 9], index=[1, 2, 3, 4, 5])
}
df_loja2 = pd.DataFrame(data_loja2)

print(df_loja1)
print("\n")
print(df_loja2)
```
