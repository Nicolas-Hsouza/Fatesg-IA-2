```python
import pandas as pd
import numpy as np
```

```python
#criando vetor

vetor= pd.Series ([1,2,3,4,5,6,7])
print(vetor)
```

```python
# Crienado um vetor com numpy e transformando para pandas.

data= np.array([1,2,3,4,5])
v = pd.Series(data)
v
```

```python
# Mudando os indices do vetor 

dado= np.array (['a','b','c','d','e'])
a = pd.Series(dado, index=[100,101,102,103,104])
a
```

```python
cubos= {'a': 0., 'b': 1., 'c': 2.}
c= pd.Series(cubos)
c 
```

```python
idades= {'joão': 25, 'Maria': 30, 'Pedro': 35, 'Ana': 25}
i= pd.Series(idades)
i
```

```python
nomes= {'a': 0., 'b': 1.,'c': 2.}
n= pd.Series(nomes, index=['b', 'c', 'a', 'y'])
n
```

```python
indices= pd.Series(5, index=[0,1,2,3,4,5])
indices
```

```python
indices2= pd.Series([1,2,3,4,5], index=['a','b','c','d','e'])
indices2
```

```python
indices3= pd.Series([1,2,3,4,5,6,7,8], index=['a','b','c','d','e','f','g','h'])

print ("Buscando os tres primeiros elementos")
print (indices2[:3])

print ("Buscando os elementos do 3° ao 6° elemento")
print (indices3 [2:6])

print ("Buscando os tres ultimos elementos")
print (indices3 [-3:0])
```
