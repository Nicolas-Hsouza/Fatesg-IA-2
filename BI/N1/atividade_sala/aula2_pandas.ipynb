```python
import pandas as pd
import numpy as np
```

```python
data = {'Name':['Tom', 'Jack', 'Steve', 'Ricky'], 'Age':[18, 34, 29, 42]}
df = pd.DataFrame(data)
print(df)
```

```python
df = pd.DataFrame(data, index=['rank1', 'rank2', 'rank3', 'rank4'])
print(df) 
```

```python
data2= [{'a':1, 'b':2,},{'a':5, 'b':10, 'c':20}]
df2 = pd.DataFrame(data2)
print(df2)
```

```python
data2= [{'a':1, 'b':2,},{'a':5, 'b':10, 'c':20}]
df2= pd.DataFrame(data2, index=['first','second'])
print (df2)
```

```python
data2= [{'a':1, 'b':2},{'a':5, 'b':10, 'c':20}]
df1= pd.DataFrame(data2, index= ['first','second'], columns=['a', 'b'])

df2= pd.DataFrame(data2, index=['first', 'second'], columns=['a', 'b1'])
print (df1)
print (df2)
```

```python
data3= {'one': pd.Series([1,2,3], index=['a', 'b', 'c']),
        'two': pd.Series([1,2,3,4], index= ['a', 'b', 'c', 'd'])}
df3= pd.DataFrame(data3)
print(df3)
```

```python
print (df3['one'])
```

```python
print("Adicionando uma nova coluna passando como series:")
df3['three']= pd.Series([10,20,30], index=['a', 'b', 'c'])
print(df3)


print("Adicionando uma nova coluna usando as colunas existentes no DataFrame: ")
df3['four']=df3['one']+df3['three']

print(df3)


```
